```python
import networkx as nx
import matplotlib.pyplot as plt

# Crear grafo dirigido
G = nx.DiGraph()

# Agregar 8 nodos
for i in range(1, 9):
    G.add_node(i)

# Cada nodo se conecta con otros 2 nodos
for i in range(1, 9):
    G.add_edge(i, (i % 8) + 1)        # siguiente nodo
    G.add_edge(i, ((i + 1) % 8) + 1)  # segundo siguiente nodo

# Dibujar
pos = nx.circular_layout(G)
nx.draw(
    G, pos,
    with_labels=True,
    node_size=2000,
    node_color="lightblue",
    arrows=True
)

plt.show()
```
